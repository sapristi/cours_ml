```python

class Perceptron:
    w0: float # biais
    w1: float
    w2: float

    def output(self, x1, x2):
        if x1 * self.w1 + x2 * self.w2 - self.w0 >= 0:
            return 1
        else:
            return 0


    def learn(self, x1, x2, t, r):
        y = self.output(x1, x2)
        if y != t:
            self.w1 = self.w1 - r * (t - y) * x1
            self.w2 = self.w2 - r * (t - y) * x2
```
